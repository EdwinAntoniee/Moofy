# Import Libraries

```python
import pandas as pd
import joblib
import os
from sklearn.feature_extraction.text import TfidfVectorizer
from sklearn.svm import LinearSVC
```

# Load Dataset

```python
train_df = pd.read_csv('../data/data_train_master.csv')
train_df.head(5)
```

# Training

## Data Splitting

```python
X_train = train_df['clean_text']
y_train = train_df['emotion']
```

## Feature Extraction & N-Gram

To better capture the contextual meaning of the text, we use both unigram and bigram features. Before feeding the data into the SVM model, the text is first transformed into numerical vectors so the model can process and learn from it effectively.

```python
vectorizer = TfidfVectorizer(max_features=10000, ngram_range=(1, 2))
X_train_vec = vectorizer.fit_transform(X_train)
```

# Training the Model

```python
svm_model = svm_model = LinearSVC(class_weight='balanced', random_state=42)
svm_model.fit(X_train_vec, y_train)
```

# Dump Model

```python
import os
import joblib

MODEL_DIR = "../models/svm"
os.makedirs(MODEL_DIR, exist_ok=True)


joblib.dump(svm_model, os.path.join(MODEL_DIR, "svm_model.joblib"))
```
